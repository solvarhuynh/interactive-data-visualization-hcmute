```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)
df = pd.DataFrame({
    'Date': pd.date_range(start='2026-01-05',
                          periods= 100, freq='D'),
    'Category': np.random.choice(['Electronics',
                                  'Clothing', 'Food'], 100),
    'Sales': np.random.randint(100, 1000, 100)
})
```

```python
print("Du lieu tho (chi tiet cao)")
print(df.head(3))
```

## Sort
giảm dần

```python
df.sort_values(by="Sales", ascending=False)
```

## Filtering

```python
df[(df['Category'] == 'Electronics') & (df['Sales'] > 1500)]
```

```python
df.loc[df['Sales'] > 500, ['Date', 'Sales']]
```

```python
df.query("Category == 'Electronics' and Sales > 500")
```

### Vẽ biểu đồ

```python
import plotly.express as px
```

```python
top10_chart = df.sort_values('Sales', ascending=False).head(10)

top10_chart['Date_label'] = top10_chart['Date'].dt.strftime('%d/%m/%Y')

fig = px.bar(
    top10_chart,
    x='Sales',
    y='Date_label',
    orientation='h',
    title='Top 10 ngày có doanh thu cao nhất',
    labels={
        'Sales': 'Doanh thu',
        'Date_label': 'Ngày'
    },
)
fig.show()
```
