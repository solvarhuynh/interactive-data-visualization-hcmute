```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)
df_orders = pd.DataFrame({
    'Order_ID': [f'ORD-{i:03d}' for i in range(1, 101)],
    'Customer_ID': np.random.choice([f'CUST-{i:02d}' for i in range(1, 21)], 100),
    'Category': np.random.choice(['Electronics', 'Clothing', 'Furniture'], 100),
    'Sales': np.random.randint(50, 500, 100)
})

agg_summary = df_orders.groupby('Category').agg(
    Total_Sales=('Sales', 'sum'),
    Avg_Sales=('Sales', 'mean'),
    Total_Orders=('Order_ID', 'count'),
    Unique_Customers=('Customer_ID', 'nunique')
).reset_index()

print(agg_summary)
```

```python
np.random.seed(42)
df_orders = pd.DataFrame({
    'Order_ID': [f'ORD-{i:03d}' for i in range(1, 6)],
    'Region': ['North', 'South', 'North', 'South', 'North'],
    'Sales': [200, 150, 300, 450, 100]
})
df_orders
```

```python
# 1. Dùng transform() để tính Tổng doanh số theo Khu vực (giữ nguyên cấp độ dòng)
df_orders['Region_Total_Sales'] = df_orders.groupby('Region')['Sales'].transform('sum')

# 2. Dùng assign() để tạo cùng lúc nhiều trường dữ liệu tỷ lệ mới (Method Chaining)
grand_total_sales = df_orders['Sales'].sum()

df_orders = df_orders.assign(
    # Tính % đóng góp trên Tổng doanh số toàn cục (Grand Total)
    Pct_Grand_Total = lambda x: (x['Sales'] / grand_total_sales) * 100,
    
    # Tính % đóng góp trong nội bộ Khu vực (Percent within Region)
    Pct_within_Region = lambda x: (x['Sales'] / x['Region_Total_Sales']) * 100
)

# Hiển thị kết quả
print(df_orders[['Order_ID', 'Region', 'Sales', 'Pct_Grand_Total', 'Pct_within_Region']])
```
