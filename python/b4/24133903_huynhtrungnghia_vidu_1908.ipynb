```python
import matplotlib.pyplot as plt
import plotly.express as px
import plotly.graph_objects as go
```

# ví dụ 1 - bieu do cot
## Matplotlib

```python
categories = ['Táo', 'Cam', 'Xoài', 'Chuối']
quantities = [20, 35, 15, 25]
colors = ['red', 'orange', 'green', 'yellow']

plt.bar(categories, quantities, color=colors)

plt.title('Số lượng trái cây bán được')
plt.xlabel('Loại trái cây')
plt.ylabel('Số lượng (kg)')

plt.show()
```

## Plotly - bieu do cot

```python
categories = ['Táo', 'Cam', 'Xoài', 'Chuối', 'Nho']
quantities = [20, 35, 15, 25, 30]

fig = px.bar(
    x=categories,
    y=quantities,
    labels={'x': 'Loại trái cây', 'y': 'Số lượng (kg)'},
    title='Số lượng trái cây bán được (Có tương tác)',
    color=categories, 
    text_auto=True)
fig.update_layout(xaxis={'categoryorder': 'total descending'})

fig.show()
```

# vd2 - bieu do line
## matplotlib - bieu do mien

```python
months = ['T1', 'T2', 'T3', 'T4', 'T5']
revenue = [10, 15, 12, 50, 8]

plt.plot(months, revenue, color='green', linewidth=2)
plt.fill_between(months, revenue, color='lightgreen', alpha=0.3)

plt.title('Biểu đồ diện tích doanh thu')
plt.xlabel('Tháng')
plt.ylabel('Doanh thu (triệu VNĐ)')
plt.grid(True)
plt.show()
```

# plotly - bieu do line

```python
months = ['T1', 'T2', 'T3', 'T4', 'T5']
revenue = [10, 15, 12, 22, 18]

fig = px.line(
    x=months,
    y=revenue,
    markers=True,
    title='Doanh thu theo tháng (Interactive Line Chart)',
    labels={'x': 'Tháng', 'y': 'Doanh thu (triệu VNĐ)'}
)


fig.update_traces(
    line_color='red',
    fillcolor='rgba(255, 0, 0, 0.3)'
)

fig.show()
```

## plotly - bieu do mien

```python
months = ['T1', 'T2', 'T3', 'T4', 'T5']
revenue = [10, 15, 12, 22, 18]

fig = px.area(
    x=months,
    y=revenue,
    title='Biểu đồ diện tích doanh thu (Interactive Area Chart)',
    labels={'x': 'Tháng', 'y': 'Doanh thu (triệu VNĐ)'},
    markers=True
)

fig.update_traces(
    line_color='red',
    fillcolor='rgba(255, 0, 0, 0.3)'
)

fig.show()
```

# vd3 - gộp 2 bieu đồ line
## plotly - graph objects

```python
months = ['T1', 'T2', 'T3', 'T4', 'T5']
rev_2023 = [10, 15, 12, 22, 18]
rev_2024 = [14, 18, 16, 28, 24]

fig = go.Figure()

fig.add_trace(go.Scatter(
    x=months,
    y=rev_2024,
    mode='lines+markers',
    name='Năm 2024',
    line=dict(color='#FFA500'),
    fill='tozeroy',                      
    fillcolor='rgba(255, 165, 0, 0.25)' 
))

fig.add_trace(go.Scatter(
    x=months,
    y=rev_2023,
    mode='lines+markers',
    name='Năm 2023',
    line=dict(color='#0000FF')
))

fig.update_layout(
    title='So sánh doanh thu 2023 vs 2024',
    xaxis_title='Tháng',
    yaxis_title='Doanh thu (triệu VNĐ)',
0    hovermode='x unified'
)

fig.show()
```

## plotly - express

```python
import plotly.express as px

months = ['T1', 'T2', 'T3', 'T4', 'T5']
rev_2023 = [10, 15, 12, 22, 18]
rev_2024 = [14, 18, 16, 28, 24]

fig = px.area(
    x=months,
    y=rev_2024,
    markers=True,
    title='So sánh doanh thu 2023 vs 2024',
    labels={'x': 'Tháng', 'y': 'Doanh thu (triệu VNĐ)'}
)

fig.data[0].name = 'Năm 2024'
fig.data[0].showlegend = True
fig.update_traces(
    line_color='#FFA500',
    fillcolor='rgba(255, 165, 0, 0.25)'
)

fig_line = px.line(
    x=months,
    y=rev_2023,
    markers=True
)
fig_line.data[0].name = 'Năm 2023'
fig_line.data[0].showlegend = True
fig_line.update_traces(line_color='#0000FF')

fig.add_trace(fig_line.data[0])

fig.update_layout(hovermode='x unified')

fig.show()
```

# vd4 - bieu do scatter
## matplotlib

```python
x = [1, 2, 3, 4, 5, 6, 7, 8, 9]
y = [5, 7, 4, 8, 6, 9, 8, 11, 10]

plt.scatter(x, y, color='#D81B60', marker= 'o', s=100)

plt.title('Biểu đồ Scatter Plot đơn giản')
plt.xlabel('Trục X')
plt.ylabel('Trục Y')
plt.grid(True)

plt.show()
```

## plotly

```python
import pandas as pd

df = pd.DataFrame({
    'Giờ học': [1, 2, 3, 4, 5, 6, 7, 8, 9],
    'Điểm thi': [45, 52, 58, 60, 68, 75, 82, 88, 95],
    'Tỷ lệ chuyên cần (%)': [60, 65, 70, 75, 80, 85, 90, 95, 100],
    'Khối ngành': ['Tự nhiên', 'Xã hội', 'Tự nhiên', 'Xã hội', 'Tự nhiên',
                   'Xã hội', 'Tự nhiên', 'Xã hội', 'Tự nhiên']
})

fig = px.scatter(
    df,
    x='Giờ học',
    y='Điểm thi',
    color='Khối ngành',
    size='Tỷ lệ chuyên cần (%)', 
    hover_data=['Tỷ lệ chuyên cần (%)'],
    title='Mối liên hệ giữa Giờ học và Điểm thi',
    size_max=20     
)

fig.show()
```

# vd5 - bieu do tron
## matplotlib

```python
labels = ['Táo', 'Cam', 'Chuối', 'Xoài']
sizes = [35, 25, 20, 20]

plt.pie(
    sizes,
    labels=labels,
    autopct='%1.1f%%', #
    startangle=90 #
)

plt.title('Tỷ lệ các loại trái cây bán được')
plt.axis('equal')  # làm bd tròn hơn  
plt.show()
```

## plotly

```python
labels = ['Táo', 'Cam', 'Chuối', 'Xoài']
sizes = [35, 25, 20, 20]

fig = px.pie(
    names=labels,
    values=sizes,
    title='Tỷ lệ trái cây bán được (Interactive Pie Chart)',
    hole = 0.4  # đục lỗ
)

fig.update_traces( #
    textinfo = 'percent+label',
    textposition = 'inside',
    pull=[0,0.115,0,0]
)
fig.show()
```
