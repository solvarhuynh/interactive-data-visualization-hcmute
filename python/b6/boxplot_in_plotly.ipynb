```python
import pandas as pd
import plotly.express as px
```

```python
df = pd.DataFrame({
    'Nhom' : ['Nhom A'] * 8  + ['Nhom B']*6,
    'Gia tri': [15, 12, 19, 20, 15, 30, 25, 31, 8, 56, 5, 6, 38, 22]
})

fig = px.box(
    df,
    x = 'Nhom',
    y = 'Gia tri',
    color = 'Nhom',
    points='all',
    hover_data=["Gia tri"],
    title="Biểu đồ Box Plot Tương tác (Plotly)"
)
fig.show()
```
