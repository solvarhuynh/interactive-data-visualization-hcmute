```python
import pandas as pd
import numpy as np
import plotly.express as px
```

```python
np.random.seed(42)

thang = [f'T{i}' for i in range(1, 13)]
chi_nhanh = ['Hà Nội', 'Đà Nẵng', 'TP.HCM']
dong_sp = ['Điện tử', 'Gia dụng', 'Thời trang']

rows = []
for t in thang:
    for cn in chi_nhanh:
        for sp in dong_sp:
            chi_phi_mkt = np.random.randint(10, 50)
            doanh_thu = np.random.randint(100, 500)
            loi_nhuan = round(doanh_thu * np.random.uniform(0.1, 0.3), 1)
            rows.append([t, cn, sp, chi_phi_mkt, doanh_thu, loi_nhuan])

df5 = pd.DataFrame(rows, columns=['Thoi_gian', 'Chi_nhanh', 'Dong_san_pham',
    'Chi_phi_MKT', 'Doanh_thu', 'Loi_nhuan'])

print('Số dòng dữ liệu:', len(df5))
df5.head()
```

```python
doanh_thu_thang = df5.groupby('Thoi_gian', sort=False)['Doanh_thu'].sum().reset_index()

doanh_thu_thang['Thoi_gian'] = pd.Categorical(doanh_thu_thang['Thoi_gian'], categories=thang, ordered=True)
doanh_thu_thang = doanh_thu_thang.sort_values('Thoi_gian')

thang_max = doanh_thu_thang.loc[doanh_thu_thang['Doanh_thu'].idxmax(), 'Thoi_gian']

mau_cot = ['red' if t == thang_max else 'lightgray'
    for t in doanh_thu_thang['Thoi_gian']]

fig5 = px.bar(doanh_thu_thang,
    x='Thoi_gian',
    y='Doanh_thu',
    text='Doanh_thu',
    title=f'Doanh thu theo tháng (Cao nhất: {thang_max})',
    labels={'Thoi_gian': 'Tháng', 'Doanh_thu': 'Doanh thu'})

fig5.update_traces(
    marker_color=mau_cot,
    textposition='outside',
    cliponaxis=False
)
fig5.update_layout(
    showlegend=False,
    margin=dict(t=90, b=60, l=60, r=30)
)

fig5.show()
```
