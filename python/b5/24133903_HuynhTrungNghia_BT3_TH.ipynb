```python
import pandas as pd
import numpy as np
import plotly.express as px
```

```python
data3 = {'Chi_phi_MKT': [10, 20, 15, 30, 25, 35, 40, 18],
    'Doanh_so': [100, 180, 130, 250, 210, 300, 320, 150],
    'Khu_vuc': ['Bắc', 'Bắc', 'Trung', 'Nam', 'Nam', 'Bắc', 'Nam', 'Trung'],
    'So_khach': [50, 80, 60, 120, 90, 150, 160, 70]}
df3 = pd.DataFrame(data3)

fig3 = px.scatter(
    df3,
    x='Chi_phi_MKT',
    y='Doanh_so',
    color='Khu_vuc',
    size='So_khach',
    trendline='ols',
    title='Tương quan giữa Chi phí tiếp thị và Doanh số',
    labels={'Chi_phi_MKT': 'Chi phí tiếp thị', 'Doanh_so': 'Doanh số'},
    size_max=30)

fig3.show()
```
