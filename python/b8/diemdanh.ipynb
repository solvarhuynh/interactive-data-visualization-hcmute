```python
import pandas as pd
import plotly.express as px

# 1. Khởi tạo dữ liệu
data = {
    'City': ['Hà Nội', 'Hải Phòng', 'Đà Nẵng', 'TP. Hồ Chí Minh', 'Cần Thơ'],
    'Lat': [21.0285, 20.8449, 16.0544, 10.8231, 10.0452],
    'Lon': [105.8542, 106.6881, 108.2022, 106.6297, 105.7469],
    'Dan_so': [8.5, 2.1, 1.2, 9.3, 1.3]
}
df = pd.DataFrame(data)

# 2. Vẽ Proportional Symbol Map tương tác
fig = px.scatter_map(
    df,
    lat="Lat",
    lon="Lon",
    size="Dan_so",      
    color="Dan_so",     
    hover_name="City",    Hiển thị tên thành phố khi di chuột
    size_max=35,         # Kích thước vòng tròn tối đa16.0, "lon": 106.5}, # Căn trung tâm Việt Nam
    map_style="carto-positron",         # Nền bản đồ sạch, đẹp
    title="Bản đồ Dân số Tương tác các Thành phố Việt Nam"
)

fig.show()
```
