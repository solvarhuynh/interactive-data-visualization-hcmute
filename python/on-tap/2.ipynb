```python
import matplotlib.pyplot as plt
import plotly.express as px
import seaborn as sns
import plotly.graph_objects as go
import pandas as pd
```

# cột

```python
categories = ['Táo', 'Cam', 'Xoài', 'Chuối']
quantities = [20, 35, 15, 25]
colors = ['red', 'orange', 'green', 'yellow']
```

```python
plt.bar(categories, quantities, color = colors)
plt.title("so luong trai cay ban dc")
plt.xlabel('danh muc')
plt.ylabel('so luong')
plt.show()
```

```python
fig = px.bar(
    x = categories,
    y = quantities,
    title = "so luong trai cay ban duoc",
    labels = {'x':'danh muc', 'y':'so luong'},
    color=colors,
    text_auto = True,
)
fig.update_layout(xaxis = {'categoryorder':'total descending'})
fig.show()
```

# line

```python
months = ['T1', 'T2', 'T3', 'T4', 'T5']
revenue = [10, 15, 12, 20, 8]
```

```python
plt.plot(months, revenue, color = 'green', linewidth = 2)
plt.fill_between(months, revenue, color = 'lightgreen', alpha = 0.3)
plt.title('bieu do dien tich doanh thu')
```
