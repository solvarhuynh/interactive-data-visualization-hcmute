```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
A = np.random.normal(loc=65, scale=8, size = 200)
B = np.random.normal(loc=75, scale=10, size = 200)
```

```python
plt.hist(A, bins=15, alpha=0.6, label = 'Lop A', color = 'blue', edgecolor = 'black')
plt.hist(B, bins=15, alpha=0.6, label = 'Lop B', color = 'red', edgecolor = 'black')

plt.title("So sanh Diem so giua lop A va lop B")
plt.xlabel('Diem so')
plt.ylabel('Tan suat')
plt.legend(loc='upper right')
plt.grid(axis='y', linestyle = '--', alpha=0.5)

plt.show()
```
