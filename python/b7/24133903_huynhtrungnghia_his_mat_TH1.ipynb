```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
np.random.seed(42)
data = np.random.normal(loc=80,scale=6,size=100)
```

```python
plt.figure(figsize=(7,4))
plt.hist(data,bins=10, color = 'blue', edgecolor='black')

plt.title('Phan bo Diem so hoc sinh')
plt.xlabel('Diem so')
plt.ylabel('Tan suat (so luong)')
plt. grid(axis='y', linestyle='--',alpha=0.7)

plt.show()
```
