```python
import plotly.express as px
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({
    "iso_alpha": ["USA", "CHN", "JPN", "DEU", "FRA", "GBR", "BRA", "AUS", "VNM"],
    "Chi_so": [85, 92, 78, 88, 70, 65, 50, 80, 75]
})

fig = px.choropleth(
    df,
    locations="iso_alpha",
    color="Chi_so",
    color_continuous_scale="YlOrRd",
    title="Chỉ số Tiêu dùng theo Quốc gia (Ví dụ minh họa)"
)

fig.show()
```

```python
df1 = pd.DataFrame({
    "iso_alpha": ["CAN", "IND", "FRA", "GBR", "KOR", "ITA", "MEX", "BRA"],
    "Chi_so": [50, 65, 45, 48, 52, 60, 55, 70]
})

fig1 = px.choropleth(
    df1,
    locations="iso_alpha",
    color="Chi_so",
    color_continuous_scale="Viridis",
    title="Chỉ số Tiêu dùng theo Quốc gia"
)

fig1.show()
```
