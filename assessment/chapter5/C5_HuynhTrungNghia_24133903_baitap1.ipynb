## Bài tập 1

```python
import pandas as pd

df_tickets = pd.DataFrame({
    'Ticket_Code': ['TCK_HN_101', 'TCK_HCM_102', 'TCK_HN_103', 'TCK_DN_104', 'TCK_HCM_105'],
    'Customer_ID': ['CUST_A', 'CUST_B', 'CUST_A', 'CUST_C', 'CUST_B']
})
df_tickets
```

### Yêu cầu 1: Trích xuất tên Thành phố sang cột City

```python
df_tickets['City'] = df_tickets['Ticket_Code'].str.split('_').str[1]
df_tickets
```

### Yêu cầu 2: Đếm số lượng khách hàng duy nhất theo từng Thành phố

```python
unique_customers = df_tickets.groupby('City')['Customer_ID'].nunique()
unique_customers
```
