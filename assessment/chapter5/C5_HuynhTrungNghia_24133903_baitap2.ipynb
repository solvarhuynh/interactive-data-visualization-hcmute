## Bài tập 2

```python
import pandas as pd

df_hotels = pd.DataFrame({
    'Booking_ID': ['BK-01', 'BK-02', 'BK-03'],
    'Check_In': pd.to_datetime(['2026-03-01', '2026-03-02', '2026-03-05']),
    'Check_Out': pd.to_datetime(['2026-03-04', '2026-03-07', '2026-03-06']),
    'Total_Fee': [300, 1000, 200]
})
df_hotels
```

### Yêu cầu 1: Tính số ngày lưu trú

```python
df_hotels['Stay_Days'] = (df_hotels['Check_Out'] - df_hotels['Check_In']).dt.days
df_hotels
```

### Yêu cầu 2: Tạo cột Daily_Rate

```python
df_hotels['Daily_Rate'] = df_hotels['Total_Fee'] / df_hotels['Stay_Days']
df_hotels
```
