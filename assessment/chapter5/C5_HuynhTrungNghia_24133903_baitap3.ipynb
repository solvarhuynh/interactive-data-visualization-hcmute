## Bài tập 3

```python
import pandas as pd

df_staff = pd.DataFrame({
    'Emp_Name': ['An', 'Bình', 'Cường', 'Dung', 'Giang'],
    'Dept': ['Sales', 'Sales', 'IT', 'IT', 'Sales'],
    'Sales': [120, 200, 150, 180, 90]
})
df_staff
```

### Yêu cầu 1: Dùng Fixed LoD tạo cột Dept_Max_Sales

```python
df_staff['Dept_Max_Sales'] = df_staff.groupby('Dept')['Sales'].transform('max')
df_staff
```

### Yêu cầu 2: Tính độ lệch doanh số cá nhân so với đỉnh phòng ban

```python
df_staff['Diff_To_Max'] = df_staff['Sales'] - df_staff['Dept_Max_Sales']
df_staff
```
