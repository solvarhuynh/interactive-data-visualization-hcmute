## Bài tập 3

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

n = 123
region = np.random.choice(['Bắc', 'Trung', 'Nam'], n)
sales = np.random.randint(100, 1000, n)

df_region = pd.DataFrame({'Region': region, 'Sales': sales})
df_region.head()
```

### 1: Tính tổng Sales theo từng Region bằng groupby()

```python
sales_by_region = df_region.groupby('Region')['Sales'].sum().reset_index()
sales_by_region
```

### 2: Tạo cột Sales_Percent

```python
tong_sales = sales_by_region['Sales'].sum()
sales_by_region['Sales_Percent'] = sales_by_region['Sales'] / tong_sales * 100
sales_by_region
```

### 3: Hiển thị bảng kết quả và vẽ Bar Chart cột đứng

```python
plt.bar(sales_by_region['Region'], sales_by_region['Sales_Percent'], color='darkorange')
plt.title('Phần trăm đóng góp doanh thu theo khu vực')
plt.xlabel('Region')
plt.ylabel('Sales_Percent (%)')
plt.tight_layout()
plt.show()
```
