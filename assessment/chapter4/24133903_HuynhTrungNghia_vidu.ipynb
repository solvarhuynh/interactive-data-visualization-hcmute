## Ví dụ 1: Pivot Table tính Subtotals & Grand Total

```python
import pandas as pd

data = {
    'Cua_Hang': ['Chi nhánh A', 'Chi nhánh A', 'Chi nhánh B', 'Chi nhánh B', 'Chi nhánh A', 'Chi nhánh B'],
    'Loai_Trai_Cay': ['Táo', 'Cam', 'Táo', 'Cam', 'Táo', 'Táo'],
    'Doanh_Thu': [150, 200, 180, 220, 130, 170]
}
df_fruits = pd.DataFrame(data)
df_fruits
```

```python
pivot = pd.pivot_table(
    df_fruits,
    index='Cua_Hang',
    columns='Loai_Trai_Cay',
    values='Doanh_Thu',
    aggfunc='sum',
    margins=True,
    margins_name='Tong_Cong'
)
pivot
```

## Ví dụ 2: Tính % đóng góp điểm số môn học

```python
import plotly.express as px

data2 = {
    'Thanh_Phan': ['Chuyên cần', 'Bài tập/Minilab', 'Giữa kỳ', 'Đồ án Môn học'],
    'Diem_So': [10.0, 8.5, 7.0, 9.0],
    'Trong_So': [10, 20, 20, 50]
}
df_diem = pd.DataFrame(data2)

df_diem['Diem_Quy_Doi'] = df_diem['Diem_So'] * df_diem['Trong_So'] / 100
tong_diem = df_diem['Diem_Quy_Doi'].sum()
df_diem['Percent_Contribution'] = df_diem['Diem_Quy_Doi'] / tong_diem * 100

print(f'Tổng điểm môn học của sinh viên: {tong_diem:.2f} / 10.0')
df_diem
```

```python
fig = px.pie(
    df_diem,
    names='Thanh_Phan',
    values='Diem_Quy_Doi',
    title='Cấu trúc điểm đóng góp của sinh viên'
)
fig.show()
```
