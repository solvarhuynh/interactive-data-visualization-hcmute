## Bài tập 1

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)

n = 50
order_id = [f'OD{i:03d}' for i in range(1, n + 1)]
category = np.random.choice(['Electronics', 'Clothing', 'Food', 'Furniture'], n)
sales = np.random.randint(50, 800, n)
profit = np.random.randint(-50, 300, n)

df = pd.DataFrame({
    'Order_ID': order_id,
    'Category': category,
    'Sales': sales,
    'Profit': profit
})
df.head()
```

### 1: Lọc các đơn hàng thuộc nhóm ngành có Sales > 300

```python
df_filtered = df[df['Sales'] > 300]
df_filtered
```

### 2: Sắp xếp kết quả theo cột Profit giảm dần

```python
df_sorted = df_filtered.sort_values(by='Profit', ascending=False)
df_sorted
```

### 3: Lấy Top 5 đơn hàng có lợi nhuận cao nhất, vẽ Bar Chart

```python
top5 = df_sorted.head(5)

plt.bar(top5['Order_ID'], top5['Profit'], color='steelblue')
plt.title('Top 5 đơn hàng có lợi nhuận cao nhất')
plt.xlabel('Order_ID')
plt.ylabel('Profit')
plt.tight_layout()
plt.show()
```
