## Bài tập 2

```python
import pandas as pd
import matplotlib.pyplot as plt

ages = [18, 22, 25, 27, 31, 35, 40, 41, 45, 52, 55, 60, 62, 65, 19, 29, 38, 48, 51, 70]
df_ages = pd.DataFrame({'Age': ages})
df_ages
```

### 1: Chia độ tuổi thành 3 nhóm bằng pd.cut

```python
bins = [17, 30, 50, 100]
labels = ['Trẻ', 'Trung niên', 'Lớn tuổi']
df_ages['Nhom_Tuoi'] = pd.cut(df_ages['Age'], bins=bins, labels=labels)
df_ages
```

### 2: Đếm số lượng khách hàng trong từng nhóm

```python
so_luong_nhom = df_ages['Nhom_Tuoi'].value_counts().sort_index()
so_luong_nhom
```

### 3: Trực quan hóa bằng Pie Chart

```python
plt.pie(so_luong_nhom, labels=so_luong_nhom.index, autopct='%1.1f%%')
plt.title('Phân bố khách hàng theo nhóm tuổi')
plt.show()
```
