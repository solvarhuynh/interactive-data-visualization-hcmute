```python
import pandas as pd

data = {
    'Thanh_Phan': ['Chuyên cần', 'Bài tập/MiniLab', 'Giữa kỳ', 'Đồ án Môn học'],
    'Diem_So': [10.0, 8.5, 7.0, 9.0],
    'Trong_So': [0.10, 0.20, 0.20, 0.50]
}
df = pd.DataFrame(data)
print(df)
```

```python
df['Diem_Quy_Doi']=df['Diem_So']*df['Trong_So']
tong_diem = df['Diem_Quy_Doi'].sum()
print(f"Tổng điểm môn học của sinh viên: {tong_diem:.2f}/10.0")
print("Bảng chi tiết % đóng góp điểm số:")
df['Percent_Contribution'] = (df['Diem_Quy_Doi']/tong_diem)*100
print(df.to_string(index=True))
```

```python
import plotly.express as px

fig = px.pie(
    names=df['Thanh_Phan'],
    values=((df['Diem_So']*df['Trong_So'])/tong_diem)*100,
    title="Biều đồ thể hiện sự đóng góp của điểm quy đổi vào tổng điểm"
)

fig.show()
```
