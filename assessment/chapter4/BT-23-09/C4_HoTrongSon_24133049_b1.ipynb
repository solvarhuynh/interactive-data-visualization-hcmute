```python
import numpy as np
import pandas as pd

np.random.seed(42)

n_rows = 10
categories = ["Electronics", "Furniture", "Office Supplies", "Clothing"]

order_ids = [f"ORD-{1000 + i}" for i in range(n_rows)]
cat_choices = np.random.choice(categories, size=n_rows)

sales = np.round(np.random.uniform(50, 1500, size=n_rows), 2)

profit_margin = np.random.uniform(-0.10, 0.30, size=n_rows)
profit = np.round(sales * profit_margin, 2)

df_transactions = pd.DataFrame(
    {
        "Order_ID": order_ids,
        "Category": cat_choices,
        "Sales": sales,
        "Profit": profit,
    }
)

print(df_transactions)
```

```python
# Câu 1

df_filtered = df_transactions.loc[df_transactions['Sales'] > 300, ['Sales']]
print(df_filtered)
```

```python
# Câu 2
df_sorted = df_transactions.sort_values(by='Profit', ascending=False)
print(df_sorted)
```

```python
# Câu 3
import plotly.express as px

top5 = df_sorted.head(5)

fig = px.bar(
    top5,
    x='Order_ID',
    y='Profit',
    labels={'Order_ID': 'Mã đơn hàng', 'Profit': 'Lợi nhuận'},
    title="Biểu đồ cột thể hiện Profit 5 đơn hàng",
    text_auto=True
)

fig.show()
```
