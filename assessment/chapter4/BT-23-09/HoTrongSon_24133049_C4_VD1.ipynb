```python
import pandas as pd

data = {
    'Cua_Hang': ['Chi nhánh A', 'Chi nhánh A', 'Chi nhánh B', 'Chi nhánh B', 'Chi nhánh A', 'Chi nhánh B'],
    'Loai_Trai_Cay': ['Táo', 'Cam', 'Táo', 'Cam', 'Táo', 'Táo'],
    'Doanh_Thu': [150, 200, 180, 220, 130, 170]
}

df_fruits = pd.DataFrame(data)
print(df_fruits)
```

```python
datatable= pd.pivot_table(
    df_fruits,
    values='Doanh_Thu',
    index='Cua_Hang',
    columns='Loai_Trai_Cay',
    aggfunc='sum',
    margins=True,
    margins_name='Tong_Cong'
)
print(datatable)
```
