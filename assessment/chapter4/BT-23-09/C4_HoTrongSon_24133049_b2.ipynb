```python
import pandas as pd

data = {
    'ages':  [18, 22, 25, 27, 31, 35, 40, 41, 45, 52, 55, 60, 62, 65, 19, 29, 38, 48, 51, 70]
}
df = pd.DataFrame(data)
print(df)
```

```python
# Câu 1
bins = [17, 30, 50, 100]
labels=['Trẻ', 'Trung niên', 'Lớn tuổi']
df['Age_Group'] = pd.cut(df['ages'],bins=bins,labels=labels)
print(df)
```

```python
# Câu 2
print(df.groupby('Age_Group')['ages'].count())
```

```python
# Câu 3
import plotly.express as px

fig = px.pie(
    names=(df.groupby('Age_Group')['ages'].count()).index,
    values=(df.groupby('Age_Group')['ages'].count()).values,
    title="Biểu đồ tròn trực quan hóa số lượng khách hàng theo từng nhóm"
)

fig.show()
```
