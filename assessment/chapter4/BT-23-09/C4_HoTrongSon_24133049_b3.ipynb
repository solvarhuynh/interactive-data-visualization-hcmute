```python
import numpy as np
import pandas as pd

np.random.seed(42)

n_samples = 123
regions = ['Miền Bắc', 'Miền Trung', 'Miền Nam']

data = {
    'Region': np.random.choice(regions, size=n_samples),
    'Sales': np.round(np.random.uniform(500, 10000, size=n_samples), 2)
}

df_sales = pd.DataFrame(data)

print(df_sales)
```

```python
# Câu 1
df_sales['regional_revenue'] = df_sales.groupby('Region')['Sales'].transform('sum')
print(df_sales)
```

```python
# Câu 2
tong_doanh_thu = df_sales['Sales'].sum()
df_sales['Sales_Percent'] = (df_sales['regional_revenue'] / tong_doanh_thu)*100
print(df_sales)
```

```python
# Câu 3
import plotly.express as px

bang_tong_hop = df_sales[['Region', 'Sales_Percent']].drop_duplicates()
print(bang_tong_hop)

fig = px.bar(
    bang_tong_hop,
    x='Region',
    y='Sales_Percent',
    color='Region',
    labels={'x': 'Vùng', 'y': 'Tỷ lệ phần trăm doanh số (%)'},
    title="Biểu đồ thể hiện tỷ lệ phần trăm doanh số theo vùng",
    text_auto=True
)

fig.show()
```
