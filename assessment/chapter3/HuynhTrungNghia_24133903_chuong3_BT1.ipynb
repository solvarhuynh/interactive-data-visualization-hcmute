```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
```

## 1. Box Plot - Trực quan hóa thu nhập
**Dataset:** https://www.kaggle.com/datasets/ruchi798/data-science-job-salaries
và đặt file là `ds_salaries.csv`

```python
df_salary = pd.read_csv('ds_salaries.csv')
df_salary.head()
```

```python
Q1 = df_salary['salary_in_usd'].quantile(0.25)
Q2 = df_salary['salary_in_usd'].quantile(0.5)
Q3 = df_salary['salary_in_usd'].quantile(0.75)
IQR = Q3 - Q1

can_duoi = Q1 - 1.5 * IQR
can_tren = Q3 + 1.5 * IQR

outliner = df_salary[(df_salary['salary_in_usd'] < can_duoi) | (df_salary['salary_in_usd'] > can_tren)]['salary_in_usd']

print('Q1:', Q1)
print('Q2 (Median):', Q2)
print('Q3:', Q3)
print('IQR:', IQR)
print('Số điểm dị biệt:', len(outliner))
print('Danh sách điểm dị biệt:', sorted(outliner.tolist()))
```

```python
fig1 = px.box(df_salary, x='salary_in_usd', points='outliers', color_discrete_sequence=['skyblue'], title='Box Plot - Thu nhập')
fig1.show()
```

## 2. Histogram - Phân tích thu nhập

```python
for so_bin in [10, 20, 50]:
    fig2 = px.histogram(df_salary, x='salary_in_usd', nbins=so_bin, color_discrete_sequence=px.colors.qualitative.Set2, title=f'Histogram thu nhập (bins={so_bin})')
    fig2.show()
```

**Lý do không dùng Bar Chart cho dữ liệu liên tục:**
Bar Chart dùng để so sánh giá trị giữa các danh mục rời rạc, mỗi cột đứng độc lập và có khoảng cách giữa các cột. Trong khi đó, thu nhập là dữ liệu liên tục, cần chia thành các khoảng để thể hiện phân phối tần suất. Histogram thể hiện đúng bản chất liên tục này, giúp thấy được hình dạng phân phối dữ liệu mà Bar Chart không làm được.

## 3. Split Violin Plot - So sánh chi tiêu giữa Nam và Nữ
**Dataset:** Tips Dataset

```python
import seaborn as sns
df_tips = sns.load_dataset('tips')
df_tips.head()
```

```python
fig3 = go.Figure()

fig3.add_trace(go.Violin(
    x=df_tips['day'][df_tips['sex'] == 'Male'],
    y=df_tips['total_bill'][df_tips['sex'] == 'Male'],
    legendgroup='Male', name='Male', side='negative',
    line_color='skyblue'
))

fig3.add_trace(go.Violin(
    x=df_tips['day'][df_tips['sex'] == 'Female'],
    y=df_tips['total_bill'][df_tips['sex'] == 'Female'],
    legendgroup='Female', name='Female', side='positive',
    line_color='deeppink'
))

fig3.update_traces(meanline_visible=True)
fig3.update_layout(violinmode='overlay', title='Split Violin Plot - Chi tiêu theo giới tính', yaxis_title='Total bill', xaxis_title='Ngày')
fig3.show()
```
