### Yêu câu 2
#### Task 2.1

```python
numeric_cols = ['Sales', 'Profit', 'Discount', 'Quantity']
corr_matrix = df_bai2[numeric_cols].corr(method='pearson')

corr_matrix
```

```python
fig_heatmap = px.imshow(
    corr_matrix,
    text_auto='.2f',
    color_continuous_scale='RdBu_r',
    zmin=-1,
    zmax=1,
    aspect='auto',
    title='Correlation Heatmap giữa các biến định lượng'
)

fig_heatmap.update_layout(
    xaxis_title='Biến',
    yaxis_title='Biến'
)

fig_heatmap.show()
```

#### Task 2.2

```python
treemap_data = (df_bai2
    .groupby(['Category', 'Sub-Category'], as_index=False)
    .agg({
        'Sales': 'sum',
        'Profit': 'sum'
    })
)

treemap_data.head()
```

```python
fig_treemap = px.treemap(
    treemap_data,
    path=['Category', 'Sub-Category'],
    values='Sales',
    color='Profit',
    color_continuous_scale='RdYlGn',
    color_continuous_midpoint=0,
    title='Hierarchical Treemap: Sales và Profit theo Category và Sub-Category'
)

fig_treemap.update_traces(textinfo='label+value+percent parent')

fig_treemap.show()
```
