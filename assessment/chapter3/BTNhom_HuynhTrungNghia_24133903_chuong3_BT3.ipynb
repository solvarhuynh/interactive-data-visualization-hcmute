# Yêu cầu 1

```python
import pandas as pd 
import numpy as np 

np.random.seed(42)

Thang = [f"Tháng {i}" for i in range(1,13)]
DoanhSo = np.random.randint(400,900, size=12)
df = pd.DataFrame({
    "Thang" : Thang,
    "DoanhSo" : DoanhSo
})

df["TyLeTangTruong"] = (df["DoanhSo"].pct_change()*100).round(2)
df["KPI_Muctieu"] = 650
list_product = ["Laptop Pro", "Smartphone X", "Tai nghe TrueWireless", "Smartwatch 5", "Máy tính bảng Air"]
df["TopProduct"] = np.random.choice(list_product, size=12)

print(df.to_string(index=False))
```

## Yêu cầu 2

```python
import plotly.express as px

# task 3.1

fig = px.bar(
    df,
    x=df["Thang"],
    y=df["DoanhSo"],
    text_auto=True,
    labels={'Thang':'Tháng','DoanhSo':'Doanh số'},
    title="Biểu đồ Doanh số và Đường tỷ lệ tăng trưởng"
)

fig.add_scatter(
    x=df["Thang"],
    y=df["TyLeTangTruong"],
    mode="lines+markers+text",
    text=df["TyLeTangTruong"],
    texttemplate="%{text:.1f}%",
    textposition="top center",
    name="Tỷ lệ tăng trưởng",
    yaxis="y2"
)

fig.update_layout(
    yaxis=dict(
        title="Doanh số"
    ),
    yaxis2=dict(
        title="Tỷ lệ tăng trưởng (%)",
        overlaying="y",
        side="right",
        ticksuffix="%"
    ),
)

# task 3.2

fig.add_hline(
    y=df["KPI_Muctieu"].iloc[0],
    line_dash="dash",
    annotation_text="Mốc KPI mục tiêu",
    annotation_position="top right"
)

fig.update_traces(
    selector=dict(type="bar"),

    textposition="inside",

    customdata=df[
        ["TyLeTangTruong", "TopProduct"]
    ].values,

    hovertemplate="""
            <b style="font-size:14px;">
                 %{x}
            </b>
            <br><br>
            Doanh số:
            <b>%{y:,.0f} triệu</b>
            <br>
            Tăng trưởng:
            <b>%{customdata[0]:.1f}%</b>
            <br>
            Top Product:
            <b>%{customdata[1]}</b>
        <extra></extra>
    """
)

fig.show()
```
